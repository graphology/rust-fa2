```python
import csv
import networkx as nx
from ipysigma import Sigma
```

```python
with open("../les-miserables.csv") as f:
    EDGES = list(csv.DictReader(f))

with open("../les-miserables-layout.csv") as f:
    NODES = list(csv.DictReader(f))
```

```python
EDGES[0], NODES[0]
```

```python
graph = nx.Graph()

for node in NODES:
    graph.add_node(node["node"], x=float(node["x"]), y=float(node["y"]))

for edge in EDGES:
    graph.add_edge(edge["source"], edge["target"], weight=float(edge["weight"]))
```

```python
Sigma(graph, node_size=graph.degree, edge_weight=None)
```
